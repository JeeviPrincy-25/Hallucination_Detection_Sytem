Your implementation directly shows these challenges:

🔴 1. Context Understanding

Model compares:

✔ Needs understanding, not keywords

🔴 2. Semantic Similarity

Example:

✔ Similar meaning → tricky

🔴 3. Ambiguity

→ incomplete information → model says Neutral

🔴 4. Logical Reasoning

Example:

🔴 5. Real-world Knowledge Gap

Example:

Model is unsure → Neutral

```python
import pandas as pd
from transformers import pipeline

# -----------------------------
# Load Dataset
# -----------------------------
df = pd.read_parquet("/content/test-00000-of-00001.parquet")


entail_df = df[df["label"] == 0].sample(5)

# -----------------------------
# Load Model (CORRECT)
# -----------------------------
classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")

# -----------------------------
# Correct Prediction Function
# -----------------------------
def predict(premise, hypothesis):

    labels = ["entailment", "contradiction"]

    result = classifier(
        hypothesis,
        candidate_labels=labels
    )

    top_label = result["labels"][0]
    score = result["scores"][0]

    if top_label == "entailment":
        return f"Correct ✅ (Confidence: {score:.2f})"
    else:
        return f"Hallucination ❌ (Confidence: {score:.2f})"

# -----------------------------
# Run Predictions
# -----------------------------
print("🔹 ENTAILMENT EXAMPLES:\n")

for i in range(len(entail_df)):
    p = entail_df.iloc[i]["premise"]
    h = entail_df.iloc[i]["hypothesis"]

    print("Premise:", p)
    print("Hypothesis:", h)
    print("Prediction:", predict(p, h))
    print("-"*50)
```
